```python
import pandas as pd
from pathlib import Path

data_path = Path.cwd().parent / "data" / "swb_segments_train_pairs.csv"
train_pairs_df = pd.read_csv(data_path)

train_pairs_df = (
    train_pairs_df.assign(
        sw_id=train_pairs_df["filename"].str.extract(r"sw(\d+)", expand=False).astype(int),
        seg_id=train_pairs_df["filename"].str.extract(r"-seg(\d+)", expand=False).astype(int),
    )
    .sort_values(["sw_id", "seg_id"])
    .drop(columns=["sw_id", "seg_id"])
    .reset_index(drop=True)
)

save_path = Path.cwd().parent / "data" / "swb_segments_train_pairs_ordered.csv"
train_pairs_df.to_csv(save_path, index=False)
train_pairs_df.head()
```
